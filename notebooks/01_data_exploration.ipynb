### 📌 Key Concept: What is a Dataset?
A **dataset** is a collection of examples used to train and test our model.
- **Training set**: Examples the model learns from (like studying for an exam)
- **Test set**: New examples used to evaluate the model (like taking the exam)

Our dataset contains **5,572 text messages** labeled as either:
- `spam` — Unwanted messages (747 examples)
- `ham` — Legitimate messages (4,825 examples)

## Step 1: Import Libraries

We need to import tools to load data, process it, and create charts.

```python
# Core libraries
import pandas as pd       # For data manipulation (like Excel in Python)
import numpy as np        # For numerical operations
import matplotlib.pyplot as plt  # For creating charts
import seaborn as sns     # For beautiful statistical plots
import os
import sys

# Add the project root to path so we can import our modules
sys.path.insert(0, os.path.join(os.getcwd(), '..'))

# Set plot style
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette('husl')

print('✅ Libraries imported successfully!')
```

## Step 2: Download and Load the Dataset

We'll download the dataset if it doesn't exist, then load it into a Pandas DataFrame.

> **What is a DataFrame?** Think of it like a spreadsheet table in Python — rows are messages, columns are attributes.

```python
from src.utils import download_dataset

# Download dataset (creates sample data if real data not available)
data_path = download_dataset('../data')

# Load the dataset
df = pd.read_csv(data_path, encoding='latin-1')

# Keep only the first two columns
df = df.iloc[:, :2]
df.columns = ['label', 'message']

# Drop any empty rows
df.dropna(inplace=True)

print(f'✅ Dataset loaded!')
print(f'   Total messages: {len(df)}')
print(f'   Columns: {list(df.columns)}')
```

## Step 3: First Look at the Data

Always start by looking at a sample of your data to understand its structure.

```python
# Show first 5 rows
print('📋 First 5 rows of the dataset:')
df.head()
```

```python
# Show basic statistics
print('📊 Dataset Statistics:')
print(f'   Shape: {df.shape} (rows, columns)')
print(f'\n   Label Distribution:')
print(df['label'].value_counts())
print(f'\n   Percentage:')
print(df['label'].value_counts(normalize=True).map('{:.1%}'.format))
```

```python
# Show example spam messages
print('🚨 Example SPAM messages:')
print('-' * 60)
for msg in df[df['label'] == 'spam']['message'].head(3).values:
    print(f'  • {msg}')
    print()

print('✅ Example HAM messages:')
print('-' * 60)
for msg in df[df['label'] == 'ham']['message'].head(3).values:
    print(f'  • {msg}')
    print()
```

## Step 4: Visualize the Data Distribution

Let's see how many spam vs ham messages we have.

```python
from src.utils import plot_label_distribution

# Plot distribution
plot_label_distribution(df)

# Notice: Ham messages are much more common than Spam
# This is called 'class imbalance' and we need to account for it!
```

## Step 5: Analyze Message Lengths

Do spam messages tend to be longer or shorter than ham messages?

```python
# Add message length column
df['message_length'] = df['message'].apply(len)
df['word_count'] = df['message'].apply(lambda x: len(x.split()))

# Statistics by label
print('📊 Message Length Statistics by Category:')
print(df.groupby('label')[['message_length', 'word_count']].describe().round(2))
```

```python
from src.utils import plot_message_length_distribution
plot_message_length_distribution(df)
```

## Step 6: Word Cloud Visualization

A **word cloud** shows the most common words. Bigger = more frequent.
Let's compare word clouds for spam vs ham.

```python
from src.utils import plot_word_cloud

# Spam word cloud
plot_word_cloud(df[df['label'] == 'spam']['message'], '🚨 SPAM - Most Common Words')

# Ham word cloud
plot_word_cloud(df[df['label'] == 'ham']['message'], '✅ HAM - Most Common Words')
```

## Step 7: Key Observations

Based on our exploration, we can observe:

1. **Class imbalance**: ~87% ham, ~13% spam — models should handle this
2. **Message length**: Spam messages tend to be longer
3. **Spam keywords**: Words like 'FREE', 'WIN', 'CALL', 'PRIZE', 'CLAIM' appear frequently in spam
4. **Ham keywords**: Words like 'OK', 'will', 'good', 'thanks' appear frequently in ham

These observations motivate our preprocessing and modeling choices!
