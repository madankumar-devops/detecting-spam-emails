### 📌 Naive Bayes Intuition

Imagine a teacher who has read 1000 emails:
- 800 are ham, 200 are spam
- The word **'FREE'** appears in 150 spam and only 5 ham emails
- The word **'meeting'** appears in 5 spam and 200 ham emails

When a new email arrives with 'FREE', the teacher thinks:
> 'I've seen FREE mostly in spam emails... this is probably SPAM!'

That's exactly how Naive Bayes works! It learns word probabilities from training data.

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os, sys

sys.path.insert(0, os.path.join(os.getcwd(), '..'))

from src.utils import download_dataset
from src.data_preprocessing import load_dataset, preprocess_dataframe
from src.feature_extraction import split_data, extract_tfidf_features
from src.models.naive_bayes import (
    train_naive_bayes, evaluate_model,
    plot_confusion_matrix, get_top_spam_words,
    predict_single_email
)

print('✅ All modules loaded!')
````

## Step 1: Load and Preprocess Data

````python
# Load data
download_dataset('../data')
df = load_dataset('../data/spam.csv')

# Preprocess
df = preprocess_dataframe(df)

# Split
X = df['cleaned_message']
y = df['label_encoded']
X_train, X_test, y_train, y_test = split_data(X, y)

# Extract TF-IDF features
X_train_tfidf, X_test_tfidf, vectorizer = extract_tfidf_features(
    X_train.tolist(), X_test.tolist()
)

print('\n✅ Data ready for training!')
````

## Step 2: Train the Naive Bayes Model

Training means the model LEARNS from the training data.
It calculates the probability of each word appearing in spam vs ham.

````python
# Train the model!
model = train_naive_bayes(X_train_tfidf, y_train)
print(f'\n🎉 Model trained in seconds!')
print(f'   This model has learned from {len(X_train)} training messages.')
````

## Step 3: Evaluate the Model

Now we test the model on emails it has NEVER SEEN BEFORE.

````python
# Evaluate on test data
metrics = evaluate_model(model, X_test_tfidf, y_test)
````

## Step 4: Confusion Matrix

The **confusion matrix** shows exactly where the model is right and wrong:

```
                 Predicted Ham  |  Predicted Spam
Actual Ham     | True Negative  |  False Positive  ← (ham marked as spam - bad!)
Actual Spam    | False Negative |  True Positive   ← (spam not caught - bad!)
```

We want maximum True Positives and True Negatives!

````python
os.makedirs('../results', exist_ok=True)
plot_confusion_matrix(
    model, X_test_tfidf, y_test,
    save_path='../results/naive_bayes_confusion.png'
)
````

## Step 5: What Did the Model Learn?

Let's see which words the model associates most strongly with SPAM.

````python
top_spam_words = get_top_spam_words(model, vectorizer, top_n=20)
````

## Step 6: Test with Your Own Emails!

Now the fun part — test the model with custom messages.

````python
# Test with custom emails
test_emails = [
    "CONGRATULATIONS! You've won a FREE iPhone! CLICK HERE to claim your prize NOW!",
    "Hey, are you free for lunch tomorrow? Let's catch up!",
    "URGENT: Your account needs verification. Send your password to confirm.",
    "Don't forget the team meeting at 3pm in Conference Room B.",
    "Earn $5000 per week working from home. No experience needed. Reply YES!",
]

print('='*60)
print('📧 SPAM DETECTION RESULTS')
print('='*60)

for email in test_emails:
    result = predict_single_email(model, vectorizer, email)
    status = '🚨 SPAM' if result['is_spam'] else '✅ HAM '
    print(f'{status} | Confidence: {result["spam_confidence"]:.1%} | Email: "{email[:50]}..."')
````
